```python
##  Processing that run in parelll
## CPU bound tasks that are heavy on CPU usage
```

```python
import multiprocessing
import time

def square_of_number():
    for i in range(5):
        time.sleep(1)
        print(f"Square : {i * i}")

def cube_number():
    for i in range(5):
        time.sleep(1)
        print(f"Cube : {i * i * i}")


# print(square_of_number())

if __name__ == "__main__":
# create 2 process
    p1 = multiprocessing.Process(target=square_of_number)
    p2 = multiprocessing.Process(target=cube_number)
    start = time.time()
    # start process

    p1.start()
    p2.start()

    # wait the process to complete
    p1.join()
    p2.join()

    total_time = time.time() - start


    print(total_time)
```
