```python
import time
import threading
import os

print(os.getcwd())
```

```python
def print_numbers():
    for i in range(5):
        time.sleep(1)
        print(f"Numbers : {i}")
def print_letters():
    for i in "abcde":
        time.sleep(1)
        print(f"letters : {i}")

start = time.time()

print_numbers()
print_letters()

total_time= time.time()- start

print(total_time)
```

```python
def print_numbers():
    for i in range(5):
        time.sleep(1)
        print(f"numbers : {i}")



def print_letters():
    for letter in "abcde":
        time.sleep(1)
        print(f"Letters : {letter}")
# create 2 thread
t1= threading.Thread(target=print_numbers)
t2 = threading.Thread(target = print_letters)

start = time.time()
#
## start the thread
t1.start()
t2.start()

## Wait for the threads to complete
t1.join()
t2.join()


# print_numbers()
# # time.sleep(2)
# print_letters()
total_time = time.time() - start
print(total_time)
```
